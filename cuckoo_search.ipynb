```python

import random
import math

# Tasks and their processing requirements
tasks = [10, 20, 15, 25, 30, 12, 18, 22]

# Fog and cloud servers with different processing speeds
servers = [("Fog1", 10), ("Fog2", 15), ("Cloud1", 25), ("Cloud2", 30)]

# Cuckoo Search parameters
N = 10             # Number of nests
iterations = 50    # Number of iterations
Pa = 0.25          # Abandonment probability
beta = 1.5         # Lévy flight parameter


# Calculate fitness of a solution
def fitness(solution):
    loads = [0] * len(servers)

    for i in range(len(tasks)):
        server = solution[i]
        loads[server] += tasks[i]

    # Calculate execution time on each server
    times = []

    for i in range(len(servers)):
        speed = servers[i][1]
        times.append(loads[i] / speed)

    # Lower execution time means better fitness
    return max(times)


# Generate a random nest (task allocation)
def create_nest():
    return [random.randint(0, len(servers) - 1)
            for _ in tasks]


# Generate a Lévy flight step using Mantegna's method
def levy_flight():
    sigma = (
        math.gamma(1 + beta) * math.sin(math.pi * beta / 2)
        / (
            math.gamma((1 + beta) / 2)
            * beta
            * 2 ** ((beta - 1) / 2)
        )
    ) ** (1 / beta)

    u = random.gauss(0, sigma)
    v = random.gauss(0, 1)

    return u / (abs(v) ** (1 / beta) + 1e-10)


# Initialize nests
nests = [create_nest() for _ in range(N)]

# Find the initial best solution
best = min(nests, key=fitness).copy()

# Main Cuckoo Search loop
for iteration in range(iterations):

    # Step 1: Generate new solutions using Lévy flight
    for i in range(N):

        new = nests[i].copy()

        step = levy_flight()

        # Convert the flight step into the number of tasks to change
        changes = max(1, min(len(tasks), int(abs(step) * 3) + 1))

        for _ in range(changes):
            task = random.randint(0, len(tasks) - 1)
            new[task] = random.randint(0, len(servers) - 1)

        # Step 2: Select the better solution
        if fitness(new) < fitness(nests[i]):
            nests[i] = new

    # Step 3: Abandon some nests and create new solutions
    for i in range(N):
        if random.random() < Pa:
            nests[i] = create_nest()

    # Step 4: Update the best solution
    current_best = min(nests, key=fitness)

    if fitness(current_best) < fitness(best):
        best = current_best.copy()

    print("Iteration:", iteration + 1,
          "| Best Fitness:", round(fitness(best), 3))


# Display final results
print("\nBEST TASK ALLOCATION")

loads = [0] * len(servers)

for i in range(len(tasks)):
    server = best[i]
    loads[server] += tasks[i]
    print("Task", i + 1, "->", servers[server][0])

print("\nSERVER LOADS")

for i in range(len(servers)):
    print(servers[i][0], "Load:", loads[i],
          "| Execution Time:", round(loads[i] / servers[i][1], 2))

print("\nBest Fitness:", round(fitness(best), 3))
```
